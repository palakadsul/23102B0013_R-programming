```r
example1 <- function(x, y)
{
  n = length(x)
  x1 = 0
  y1 = 0
  z1 = 0

  for (i in 1:n)
  {
    x1[i] = x[i]^2
    y1[i] = y[i]^2
    z1[i] = (x[i]/y[i])^2
  }

  sum_square_x = sum(x1)
  sum_square_y = sum(y1)
  sum_square_z = sum(z1)

  g = sum_square_x/sum_square_y
  h = sum_square_z

  cat("The value of g and h are", g,
      "respectively", "\n")
}
```

```r
x = c(10, 20, 30)
y = c(1, 2, 3)

example1(x, y)

x = c(67, 87, 26, 85, 6, 45)
y = c(54, 64, 22, 94, 20, 88)

example1(x, y)
```
